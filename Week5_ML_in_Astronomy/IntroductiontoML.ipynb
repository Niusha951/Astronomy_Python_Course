## What is Machine Learning?

Machine Learning is a field of computer science that focuses on building algorithms that can **learn patterns from data** and make predictions or decisions without being explicitly programmed for each task.

## When Do We Use Machine Learning?

Machine learning becomes most useful when a problem is too complex to solve with a simple rule or formula or when the data is too large or unstructured to handle manually.
Here are two common situations where ML shines:
1. Tasks that are hard to program directly
    * The problem doesn't have a clear step-by-step solution.
    * The data is high-dimensional, messy, or nonlinear (e.g., images, spectra, time series).
2. Tasks that require adaptation
    * The model needs to improve over time or respond to new information (e.g., weather forecasting, galaxy classification, spam filtering).

However, remember that ML is not magic, it's only as good as the data and assumptions that go into it. We'll also discuss where ML can fail and how to recognize its limits.

## The (Formal) Structure of the Learning Problem

A **Learning Problem** in Machine Learning can be broken down into a few key parts:
* **X** – The **instance space** or domain set of examples. Each example is usually represented as an n-dimensional vector. The components of these vectors are called _features_.

_Example:_ In astronomy, a galaxy might be described by its brightness, color, and size.

* **Y** – The **label set**, which represents the possible outputs for each element in X. In **supervised learning**, these are the outcomes we want to predict.

_Example:_ The galaxy type (spiral, elliptical, irregular).

* **S** – The **training set**, a collection of observed pairs of features and labels: $(x,y)$.


We often assume that there is some true function $f:X→Y$ that describes how features map to labels. Problems where this function exists and we try to approximate it are called **supervised learning** problems.

*  **h** – The **hypothesis** or output of a learning algorithm. This is our model, which tries to approximate the true function: $h:X→Y$.

_Important:_ $h$ is almost never perfect, so our goal is to make it as close to $f$ as possible.

In short: the learning problem is about **learning a function from examples**. We have X (inputs), Y (outputs), and S (training data). The goal is to approximate the true function f with a model h.

## Loss Functions

To measure how well our model $h$ performs, we use a **loss function**.

* A **loss function** quantifies how "wrong" our predictions are. Large loss = very wrong, small loss = close to correct.
* Common examples:
    * **Mean Squared Error (MSE)**: Measures the average squared difference between predicted and true values.

    $$
    L_S(h) = \frac{1}{m} \Sigma |h(x) - y(x)|^2
    $$
    
    * **Absolute Error**: Measures the average absolute difference.
    * **0-1 Loss**: Used in binary classification; 0 if prediction is correct, 1 if incorrect.


The loss function guides the learning algorithm. By minimizing it, we improve our predictions.



## No Free Lunch Theorem

We might hope for a "perfect" learning algorithm that works for any problem, but **this is impossible**.

The **No Free Lunch Theorem** states that no algorithm can perform well on every possible dataset. An algorithm that does well on one set of problems may fail completely on others.

_Lesson:_ ML methods must be chosen carefully depending on the problem and dataset.

## Why is Learning Difficult?

Learning is challenging because:
* The true function $f$ can be complex or highly nonlinear.
* We usually do **not know the underlying distribution** of our data. The training set is a sample from this unknown distribution.

This means we need to carefully consider the uncertainty in our model predictions.

## Uncertainty in Machine Learning

Errors in our model can be thought of as having two components:
 1. **Approximation error** ($ϵ_{app}$) – How far our chosen model $h$ is from the true function $f$.
 2. **Estimation error** ($ϵ_{est}$) – Error due to limited or noisy training data.

 
Together, these make up the total error in our predictions. Because the data is random, the error itself is a random variable.

## Bias-Variance Tradeoff

To reduce errors, we might try using a richer set of models:
* More complex models → smaller approximation error ($ϵ_{app}$), because they can fit the true function better.
* But more complex models → larger estimation error ($ϵ_{est}$), because they can overfit the training data (high variance).

This balance is called the **bias-variance tradeoff**.
* Low bias → model can fit the data well
* Low variance → model predictions are stable across different datasets

In statistics, this tradeoff has formal limits, e.g., the Cramer-Rao bound.

## Probably Approximately Correct (PAC) Learning

ML errors can also be formalized using PAC learning:
* ϵ → error due to the learner (approximation error)
* δ → error due to sampling from a limited dataset (estimation error)

PAC learning captures the idea that:
    With high probability ($1−δ$), our learned function $h$ will be approximately correct (error $≤ ϵ$).

This framework lets us **quantify how much data we need** for learning, depending on the complexity of the problem.

## Worry About the Data!

Even the best ML methods cannot fix **biased training data**.
* Sophisticated algorithms reduce learner error ($ϵ$), but
* If the training data is non-representative, the learned model will also be biased.

Always check your dataset before trusting ML predictions!
